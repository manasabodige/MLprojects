```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("⏳ Step 1: Downloading Movie Dataset...")
url = "https://raw.githubusercontent.com/YBI-Foundation/Dataset/main/Movies%20Recommendation.csv"
df = pd.read_csv(url)

print("⏳ Step 2: Selecting text features for recommendation...")
# Combine description features: Genre, Keywords, Tagline, Cast, and Director
features = df[['Movie_Genre', 'Movie_Keywords', 'Movie_Tagline', 'Movie_Cast', 'Movie_Director']].fillna('')
combined_features = features['Movie_Genre'] + ' ' + features['Movie_Keywords'] + ' ' + features['Movie_Tagline'] + ' ' + features['Movie_Cast'] + ' ' + features['Movie_Director']

print("⏳ Step 3: Computing Similarity Matrices using TF-IDF and Cosine Similarity...")
vectorizer = TfidfVectorizer(stop_words='english')
feature_vectors = vectorizer.fit_transform(combined_features)
similarity_matrix = cosine_similarity(feature_vectors)

# Recommendation function
def recommend_movies(favorite_movie, num_recommendations=5):
    try:
        # Find the closest match to the user's movie choice
        movie_idx = df[df['Movie_Title'].str.contains(favorite_movie, case=False, na=False)].index[0]
        similarity_scores = list(enumerate(similarity_matrix[movie_idx]))
        sorted_similar_movies = sorted(similarity_scores, key=lambda x: x[1], reverse=True)[1:num_recommendations+1]

        print(f"\n🎬 Top {num_recommendations} movie recommendations for fans of '{df.iloc[movie_idx]['Movie_Title']}':\n")
        for i, movie in enumerate(sorted_similar_movies, 1):
            print(f"{i}. {df.iloc[movie[0]]['Movie_Title']}")
    except IndexError:
        print(f"\n❌ Movie '{favorite_movie}' not found in the dataset. Try another title like 'Avatar' or 'Iron Man'.")

print("\n✅ PROJECT 3 COMPLETE! TESTING THE RECOMMENDATION SYSTEM:")
# Run a quick test case recommendation
recommend_movies("Avatar")
```
